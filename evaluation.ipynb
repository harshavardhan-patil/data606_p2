```python
import pandas as pd
import numpy as np
from scipy.sparse import csr_matrix, lil_matrix
from sklearn.neighbors import NearestNeighbors
from sklearn.model_selection import train_test_split
from pathlib import Path
from tqdm import tqdm
from sklearn.decomposition import TruncatedSVD
from collaborative_filtering import run_cf_with_svd, evaluate_model

DATA_DIR = Path('data/')

ratings_df = pd.read_csv(DATA_DIR / 'Ratings.csv')
```

```python
k_values = [5, 10, 15, 20, 50, 100]
results = {}

train_matrix_csr, test_matrix, test_df, user_to_index, book_to_index = run_cf_with_svd(ratings_df)
for k in k_values:
    mad = evaluate_model(train_matrix_csr, test_matrix, test_df, user_to_index, book_to_index, k_neighbors=k)
    print(f"k={k}, MAD={mad:.4f}")
```

```python
sample_ratios = [0.6, 0.65, 0.7, 0.75, 0.8, 0.85, 0.9]
results = {}

for ratio in sample_ratios:
    test_size = 1 - ratio
    train_matrix_csr, test_matrix, test_df, user_to_index, book_to_index = run_cf_with_svd(ratings_df, test_size=test_size)
    mad = evaluate_model(train_matrix_csr, test_matrix, test_df, user_to_index, book_to_index, k_neighbors=5)
    print(f"ratio={ratio:.2f}, MAD={mad:.4f}")
```
